## Loan Data Cleaning & Preparation with NumPy

### Project Overview

This project focuses on cleaning, preprocessing, and preparing a real-world loan dataset from the LendingClub platform using primarily **NumPy** operations and functionalities. The objective is to demonstrate strong proficiency in low-level data manipulation techniques without relying heavily on higher-level libraries such as pandas.

The primary goal of this stage of the project is to transform raw and incomplete financial data into a structured and machine-learning-ready dataset that can later be used for predictive modeling and risk analysis tasks.

Through this workflow, the project showcases:

- Efficient numerical data handling with NumPy
- Detection and treatment of missing values
- Data type management
- Preparation of structured arrays for downstream ML applications

### Splitting the Dataset

#### Splitting the Columns

To simplify the preprocessing workflow, we separate the dataset into:

- Columns containing string (categorical) data
- Columns containing numerical data

This separation is performed by identifying which columns produced `NaN` values when calculating the column means. Columns with `NaN` means correspond to non-numerical features, while the remaining columns are classified as numerical.

```python
columns_strings = np.argwhere(np.isnan(temporary_mean)).squeeze()
columns_strings
```

```python
columns_numeric = np.argwhere(np.isnan(temporary_mean) == False).squeeze()
columns_numeric
```

- `loan_data_strings` → contains all categorical/string variables

```python
loan_data_numeric = np.genfromtxt("loan-data.csv", 
                                  delimiter = ";", 
                                  skip_header = 1, 
                                  autostrip = True, 
                                  usecols = columns_numeric, 
                                  filling_values = temporary_fill)
loan_data_numeric
```

#### Checkpoint Function

A custom checkpoint function is created to save intermediate preprocessing results.

The function stores:

- The dataset headers
- The processed data arrays

using NumPy’s `.npz` format.

```python
def checkpoint(file_name, checkpoint_header, checkpoint_data):
    np.savez(file_name, header = checkpoint_header, data = checkpoint_data)
    checkpoint_variable = np.load(file_name + ".npz")
    return checkpoint_variable
```

##### Filling Sub Grade

Some observations contain a missing `sub_grade` value while still having an associated `grade`.

To preserve information:

- Missing sub-grades are filled using the corresponding grade category
- The lowest sub-grade within each grade group (e.g. A5, B5) is assigned as a placeholder value

```python
for i in np.unique(loan_data_strings[:,3])[1:]:
    loan_data_strings[:,4] = np.where((loan_data_strings[:,4] == '') & (loan_data_strings[:,3] == i),
                                      i + '5',
                                      loan_data_strings[:,4])
```

```python
np.unique(loan_data_strings[:,4], return_counts=True)
```

A small number of observations contain missing values for both `grade` and `sub_grade`. These are temporarily assigned the category `H1` to maintain consistency and avoid null entries.

```python
loan_data_strings[:,4] = np.where(loan_data_strings[:,4] == '',
                                  'H1',
                                  loan_data_strings[:,4])
```

```python
np.unique(loan_data_strings[:,4], return_counts=True)
```

##### Converting Sub Grade

After cleaning the `sub_grade` column, the categorical labels are converted into numerical representations for easier implementation in future workflows.

A dictionary mapping is created to assign a unique integer to each sub-grade category. For example:

- `A1` → 1

To reduce categorical complexity and create a more compact feature, we group individual states into broader U.S. geographic regions based on U.S. Census classifications.

#### Checkpoint 2: Numeric

##### Exchange Rate

To prepare the dataset for currency standardization, historical EUR/USD exchange rate data is imported from our CSV file.

Only the adjusted closing price column is used, since it contains the relevant monthly exchange rate values required for the conversion process.

```python
EUR_USD = np.genfromtxt("EUR-USD.csv", delimiter = ",", autostrip=True, skip_header=1, usecols=3)
EUR_USD
```

The previously cleaned `issue_date` column is used to associate each loan with its corresponding monthly EUR/USD exchange rate.

For each observation:

- The loan issue month is matched to the appropriate exchange rate
- Missing issue dates are replaced using the average exchange rate value

```python
loan_data_strings[:,0]
```

```python
exchange_rate = loan_data_strings[:,0]

for i in range(1,13):
    exchange_rate = np.where(exchange_rate == i,
                             EUR_USD[i-1],
                             exchange_rate)

exchange_rate = np.where(exchange_rate == 0,
                         np.mean(EUR_USD),
                         exchange_rate)

exchange_rate
```

The exchange rate values are reshaped and appended to the numerical dataset as a new column named `exchange_rate`.

```python
loan_data_numeric.shape
```

```python
exchange_rate = np.reshape(exchange_rate, (10000,1))
```

```python
loan_data_numeric = np.hstack((loan_data_numeric, exchange_rate))
```

```python
header_numeric = np.concatenate((header_numeric, np.array(["exchange_rate"])))
header_numeric
```

##### From USD to EUR

Several financial variables in the dataset are originally expressed in U.S. dollars.

Using the previously added `exchange_rate` feature, the following columns are converted from USD to EUR:

- `loan_amnt`
- `funded_amnt`
- `installment`
- `total_pymnt`

The conversion is performed by dividing each value by its corresponding monthly exchange rate.

The converted variables are then appended to the numerical dataset as additional columns, allowing both the original USD values and standardized EUR values to remain available for future analysis.

```python
header_numeric
```

```python
columns_dollar = np.array([1,2,4,5])
```

```python
loan_data_numeric[:,6]
```

```python
for i in columns_dollar:
    loan_data_numeric = np.hstack((loan_data_numeric, np.reshape(loan_data_numeric[:,i] / loan_data_numeric[:,6], (10000,1))))
```

```python
loan_data_numeric
```
